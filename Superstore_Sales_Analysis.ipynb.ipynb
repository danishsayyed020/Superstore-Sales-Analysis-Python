```python
import pandas as pd

file_path = r"C:\Users\Danish\Downloads\Superstore.csv"

df = pd.read_csv(file_path, encoding="cp1252")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

df.info()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
```

```python
profit_margin = (total_profit / total_sales) * 100

print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Check dataset information
print("\nDataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# Business KPIs

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
total_products = df["Product ID"].nunique()

profit_margin = (total_profit / total_sales) * 100

print("========== SUPERSTORE BUSINESS REPORT ==========")
print("Total Sales: $", round(total_sales, 2))
print("Total Profit: $", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Unique Products:", total_products)
print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
# Category-wise Sales and Profit Analysis

category_analysis = df.groupby("Category").agg({
    "Sales": "sum",
    "Profit": "sum",
    "Quantity": "sum"
}).round(2)

print(category_analysis)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Prepare category-wise sales data
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

# Create bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Total Sales by Product Category", fontsize=16)
plt.xlabel("Product Category")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Group sales and profit by category

category_performance = df.groupby("Category")[["Sales", "Profit"]].sum()

print(category_performance.round(2))
```

```python
import matplotlib.pyplot as plt

category_performance.plot(
    kind="bar",
    figsize=(10, 6),
    width=0.8
)

plt.title("Sales vs Profit by Product Category", fontsize=15)
plt.xlabel("Product Category")
plt.ylabel("Amount ($)")
plt.xticks(rotation=0)
plt.legend(["Sales", "Profit"])
plt.tight_layout()
plt.show()
```

```python
# Region-wise Sales and Profit Analysis

region_analysis = df.groupby("Region").agg({
    "Sales": "sum",
    "Profit": "sum",
    "Order ID": "nunique"
}).round(2)

region_analysis = region_analysis.sort_values(
    by="Sales", ascending=False
)

print(region_analysis)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

region_sales = df.groupby("Region")["Sales"].sum()
region_sales = region_sales.sort_values(ascending=False)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=region_sales.index,
    y=region_sales.values
)

plt.title("Total Sales by Region", fontsize=15)
plt.xlabel("Region")
plt.ylabel("Total Sales ($)")
plt.tight_layout()
plt.show()
```

```python
# Discount vs Profit Analysis

discount_analysis = df.groupby("Discount").agg({
    "Sales": "sum",
    "Profit": "sum",
    "Order ID": "count"
}).round(2)

discount_analysis = discount_analysis.rename(
    columns={"Order ID": "Number of Records"}
)

print(discount_analysis)
```

```python
# Convert Order Date to datetime format

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)

print(df["Order Date"].dtype)
```

```python
# Extract month and year from Order Date

df["Year-Month"] = df["Order Date"].dt.to_period("M")

# Calculate monthly sales

monthly_sales = df.groupby("Year-Month")["Sales"].sum()

print(monthly_sales)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o",
    linewidth=2
)

plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=90)
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Top 10 Most Profitable Products

top_products = (
    df.groupby("Product Name")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_products)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Most Profitable Products")
plt.xlabel("Total Profit ($)")
plt.ylabel("Product Name")

plt.tight_layout()
plt.show()
```

```python
# FINAL BUSINESS INSIGHTS

# Best-performing category by sales
best_category = df.groupby("Category")["Sales"].sum().idxmax()

# Most profitable category
profitable_category = df.groupby("Category")["Profit"].sum().idxmax()

# Best-performing region by sales
best_region = df.groupby("Region")["Sales"].sum().idxmax()

# Best month by sales
best_month = df.groupby("Year-Month")["Sales"].sum().idxmax()

# Most profitable product
best_product = df.groupby("Product Name")["Profit"].sum().idxmax()

print("========== FINAL BUSINESS INSIGHTS ==========")
print("Highest Sales Category:", best_category)
print("Most Profitable Category:", profitable_category)
print("Highest Sales Region:", best_region)
print("Highest Sales Month:", best_month)
print("Most Profitable Product:", best_product)
```
